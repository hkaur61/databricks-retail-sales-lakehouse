```python
sales_enriched = spark.table("retail_catalog.retail_silver.sales")
```

```python
from pyspark.sql import functions as F

#Create the fact table
fact_sales = sales_enriched.select(
    "order_id",
    "customer_id",
    "product_id",
    "order_date",
    "quantity",
    "unit_price",
    "sales_amount"
)
```

```python
fact_sales.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_catalog.retail_gold.fact_sales")
```

```python
#display(fact_sales)
```

```python
#STEP 10 — Create Gold aggregations
daily_sales = (
    fact_sales
    .groupBy("order_date")
    .agg(
        F.sum("sales_amount").alias("total_sales"),
        F.sum("quantity").alias("total_units"),
        F.countDistinct("order_id").alias("total_orders")
    )
)
```

```python
#display(daily_sales)
```

```python
daily_sales.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_catalog.retail_gold.daily_sales")
```
