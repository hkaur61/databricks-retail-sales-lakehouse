```python
#Read Bronze
sales_df = spark.table("retail_catalog.retail_bronze.sales")
customers_df = spark.table("retail_catalog.retail_bronze.customers")
products_df = spark.table("retail_catalog.retail_bronze.products")
```

```python
sales_enriched = spark.table("retail_catalog.retail_silver.sales")
```

```python
#Add data quality checks

from pyspark.sql import functions as F

null_customer_count = (
    sales_df
    .filter(F.col("customer_id").isNull())
    .count()
)

negative_quantity_count = (
    sales_df
    .filter(F.col("quantity") <= 0)
    .count()
)

duplicate_order_count = (
    sales_df
    .groupBy("order_id")
    .count()
    .filter(F.col("count") > 1)
    .count()
)
```

```python
assert null_customer_count == 0
assert negative_quantity_count == 0
assert duplicate_order_count == 0
```

```python
#validation and data quality checks
invalid_records = sales_enriched.filter(
    (F.col("customer_id").isNull()) |
    (F.col("product_id").isNull()) |
    (F.col("quantity") <= 0)
)

print("Invalid records:", invalid_records.count())
```

```python
null_orders = sales_enriched.filter(
    F.col("order_id").isNull()
).count()

duplicate_orders = (
    sales_enriched
    .groupBy("order_id")
    .count()
    .filter(F.col("count") > 1)
    .count()
)
```
