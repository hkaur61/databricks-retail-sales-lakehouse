```python
from pyspark.sql import functions as F

customers_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("/Volumes/retail_catalog/retail_schema/retail_data/raw_data/Customers.csv")
)
```

```python

customers_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_catalog.retail_bronze.customers")
```

```python
display(customers_df)
```
